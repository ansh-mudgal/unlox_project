```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("/content/student_marks.csv")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df["Exam_Marks"]=df["Exam_Marks"].fillna(df["Exam_Marks"].mean())
```

```python
df["Study_Hours"]=df["Study_Hours"].fillna(df["Study_Hours"].mean())
```

```python
df.isnull().sum()
```

```python
#outlier Treatment
sns.boxenplot(df[["Exam_Marks","Study_Hours"]])
plt.title("outlier detection")
plt.show()
```

```python
#outlier treatemnt
for col in ["Exam_Marks","Study_Hours"]:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[col]=np.where(df[col]>upper,upper,np.where(df[col]<lower,lower,df[col]))
```

```python
#outlier Treatment
sns.boxenplot(df[["Exam_Marks","Study_Hours"]])
plt.title("outlier detection")
plt.show()
```

```python
#universe analysis
sns.distplot(df["Exam_Marks"])

plt.show()
```

```python
#universe analysis
sns.distplot(df["Study_Hours"])

plt.show()
```

```python
#outlier Treatment
sns.regplot(x="Exam_Marks",y="Study_Hours",data=df)
plt.show()
#
```
